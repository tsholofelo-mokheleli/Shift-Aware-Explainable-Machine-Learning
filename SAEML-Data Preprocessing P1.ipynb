### **2017**

```python
mh2017 = pd.read_csv('OSMI Mental Health in Tech Survey 2017.csv', encoding='latin1')
mh2017 = mh2017.drop(columns=columns_to_remove, errors='ignore')

patterns = ['', '*','.1','', '', '', '','_','','', '', '']
for pattern in patterns:
    mh2017.columns = mh2017.columns.str.replace(pattern, '', regex=False)

mh2017.columns = mh2017.columns.str.strip()

mh2017.columns = (
    mh2017.columns
    .str.replace('Â', '', regex=False)              # Remove 'Â'
    .str.replace('\s+', ' ', regex=True)            # Replace multiple spaces with a single space
    .str.strip()                                    # Remove leading/trailing spaces
    .str.replace('</?strong>', '', regex=True)      # Remove <strong> and </strong> tags
    .str.replace('</?em>','',regex=True)            # Remove <em> and </em> tags
)

mh2017.shape
```

```python
# show all columns without truncating the columns
# pd.set_option('display.max_columns', None)
# print(mh2017.columns)
```

```python
mh2017.columns = mh2017.columns.str.replace(
    "If they knew you suffered from a mental health disorder, how do you think that team members/co-workers would react?",
    "If they knew you suffered from a mental health disorder, how do you think that your team members/co-workers would react?",
    regex=False
)
```

```python
mh2017['If yes, what condition(s) have you been diagnosed with?'] = mh2017[columns_to_combine].apply(
    lambda row: ', '.join(row.dropna().astype(str)), axis=1
)
mh2017.drop(columns=columns_to_combine, inplace=True)
```

```python
intersection = set(columns_to_keep).intersection(mh2017.columns)
mh2017 = mh2017.drop(columns=set(mh2017.columns) - intersection)

mh2017.shape
```

```python
print(mh2017.info())
```
