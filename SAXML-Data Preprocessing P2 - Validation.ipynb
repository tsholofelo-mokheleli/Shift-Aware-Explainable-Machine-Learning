```python
# Define the acceptable age bounds
MIN_AGE = 18
MAX_AGE = 80

# Use boolean indexing to identify rows where 'Age' is outside the range
out_of_range_mask = (mh_dataset['Age'] < MIN_AGE) | (mh_dataset['Age'] > MAX_AGE)

# Set the 'Age' value in these identified rows to NaN
mh_dataset.loc[out_of_range_mask, 'Age'] = np.nan

print(f"Highest age: {mh_dataset['Age'].max()}")
print(f"Lowest age: {mh_dataset['Age'].min()}")
```
